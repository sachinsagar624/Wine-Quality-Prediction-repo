outliers

```python
q1=df["residual sugar"].quantile(0.25)
q3=df["residual sugar"].quantile(0.75)
iqr1=q3-q1
lr=q1-1.5*iqr1
ur=q3+1.5*iqr1
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr1)
print("Lower Bound:", lr)
print("Upper Bound:", ur)
```

```python
print("Outliers:", ((df["residual sugar"] < lr) | 
                   (df["residual sugar"] > ur)).sum())
```

```python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    
    print(col, ":", outliers)
```

///////////////////////////////////////--------      AB HUM MODEL TRAIN KRENGE 

```python
x= df.drop("quality", axis=1)
y = df["quality"]

print("X shape:", x.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print("X_train:", x_train.shape)
print("X_test :", x_test.shape)
print("Y_train:", y_train.shape)
print("Y_test :", y_test.shape)
#"Stratify=y is used to maintain the same class distribution in the training and testing datasets, especially when the target variable is imbalanced."
```

//////////////////--------------Feature Scaling.

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
x_train_scaled = scalar.fit_transform(x_train)
x_test_scaled = scalar.transform(x_test)

print("X_train_scaled shape:", x_train_scaled.shape)
print("X_test_scaled shape :", x_test_scaled.shape)
```

Chalo 👍 ab first ML model — Logistic Regression.

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression(max_iter=1000)
lr.fit(x_train_scaled,y_train)
```

```python
print("Model trained successfully")
```

```python
y_pred = lr.predict(x_test_scaled)
print(y_pred[:20])
```

Accuracy + Classification Report

```python
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(y_test, y_pred)*100
print(accuracy)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print("Logistic Regression Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

# Project Summary

This project focuses on predicting wine quality using Machine Learning classification algorithms.

### Dataset Summary

* Original records: 6,497
* Duplicate records removed: 1,179
* Final records: 5,318
* Input features: 11
* Target variable: `quality`
* Missing values: None
* Train set: 4,254 records
* Test set: 1,064 records

### Machine Learning Models

The following classification models were trained and evaluated:

* Logistic Regression
* Decision Tree
* Random Forest
* Balanced Random Forest
* Tuned Random Forest
* K-Nearest Neighbors (KNN)
* Support Vector Machine (SVM)

### Model Performance

| Model                  | Test Accuracy |
| ---------------------- | ------------: |
| Random Forest          |        57.33% |
| Tuned Random Forest    |        56.20% |
| Balanced Random Forest |        55.17% |
| SVM                    |        53.85% |
| Logistic Regression    |        53.38% |
| KNN                    |        50.66% |
| Decision Tree          |        45.11% |

The Random Forest classifier achieved the highest test accuracy among the evaluated models.
Note: The tuned Random Forest had a slightly lower test accuracy (56.20%), so hyperparameter tuning did not improve the held-out test result in this experiment.

### Key Findings

* The target variable was highly imbalanced.
* Wine quality classes 5 and 6 contained the majority of observations.
* Rare classes such as 3, 4, and 9 were difficult for the models to predict.
* Alcohol was the most important feature according to Random Forest feature importance.
* Other influential features included density, volatile acidity, total sulfur dioxide, and chlorides.

### Final Prediction

The saved Random Forest model was successfully used to predict the quality of a new wine sample.

**Predicted Wine Quality: 6**

Final Conclusion

The project successfully demonstrated an end-to-end Wine Quality Classification workflow:

Data Cleaning → EDA → Duplicate Removal → Feature/Target Separation → Train-Test Split → Scaling → Multiple ML Models → Model Evaluation → Confusion Matrix → Feature Importance → Model Saving → New Wine Prediction

The Random Forest classifier achieved 57.33% test accuracy, the highest among the tested models. However, the low macro F1-score and poor performance on rare quality classes show that the model is affected by the dataset's class imbalance.


## Conclusion

In this project, a Wine Quality Prediction system was developed using Machine Learning classification techniques. The dataset initially contained 6,497 records, and 1,179 duplicate records were removed, resulting in 5,318 unique records with 11 input features.

The data was explored and analyzed to understand the distribution of wine quality and the relationship between different chemical properties and wine quality. The analysis showed that alcohol had the strongest positive correlation with wine quality, while density, volatile acidity, and chlorides showed negative relationships.

Multiple Machine Learning algorithms were trained and evaluated, including Logistic Regression, Decision Tree, Random Forest, Balanced Random Forest, KNN, SVM, and a tuned Random Forest model. Among the tested models, the Random Forest classifier achieved the highest test accuracy of approximately 57.33%.

The model showed better performance for the majority classes, particularly quality levels 5 and 6. However, the model had difficulty identifying rare quality classes such as 3, 4, and 9 because of the highly imbalanced target distribution.

Feature importance analysis showed that alcohol, density, volatile acidity, total sulfur dioxide, and chlorides were among the most influential features used by the Random Forest model.

Finally, the trained Random Forest model was saved using Joblib and successfully used to predict the quality of a new wine sample. The sample used in the project was predicted to have a wine quality score of 6.

Overall, this project demonstrates a complete Machine Learning classification workflow, including data cleaning, exploratory data analysis, preprocessing, model training, model evaluation, feature importance analysis, model saving, and prediction on new data.
